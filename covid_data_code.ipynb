```python
from pyspark.sql import SparkSession

# Создаём подключение к Spark кластеру
spark = SparkSession.builder \
    .appName("CovidAnalysis") \
    .master("spark://spark-master:7077") \
    .getOrCreate()

print("✅ Spark подключен!")
print(f"Версия: {spark.version}")
```

```python
# Читаем файл как простой текст
text_rdd = spark.sparkContext.textFile("file:///home/jovyan/work/covid-data.csv")

# Показываем первые 3 строки
print("Первые 3 строки файла:")
for line in text_rdd.take(3):
    print(line)
```
